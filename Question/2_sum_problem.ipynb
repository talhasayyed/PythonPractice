# Two Sum Problem

you have given 
```
arr = [2, 4, 3, 5, 7, 8, 1, 6]
targets = [5, 8, 7]
```

For each target number (5, 8, 7), find which two numbers from the array sum up to the target.

````python

# index
#    # 0, 1, 2, 3, 4, 5, 6, 7
arr = [2, 4, 3, 5, 7, 8, 1, 6]
targets = [5, 8, 7]
print('input loaded!')
````

---

### Bruteforce way 

````python
def find_two_sum_bruteforce(arr, target):
    pairs = []
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            arr_i = arr[i]
            arr_j = arr[j]
            if arr_i + arr_j == target:
                pairs.append((arr_i, arr_j))
    return pairs

for t in targets:
    pairs = find_two_sum_bruteforce(arr, t)
    print(f"{t} -> {pairs}")
````

#### Time Space complexity

Outer loop → runs n times
Inner loop → runs (n - i - 1) times

Total comparisons:
`(n-1) + (n-2) + (n-3) + ... + 1`

This equals:
n(n-1)/2  →  O(n²)

Final Time Complexity:
O(n²)

----

### Optimised way

````python
def find_all_pairs(arr, target):
    seen = set()
    pairs = set()
    
    for num in arr:
        complement = target - num
        
        if complement in seen:
            # sort to avoid duplicate like (7,1) and (1,7)
            pair = tuple(sorted((num, complement)))
            pairs.add(pair)
        
        seen.add(num)
    
    return list(pairs)

for target in targets:
    print(f'{target} -> {find_all_pairs(arr, target)}')
````

#### ⏱ Time Complexity

We loop only once:
`for num in arr → runs n times`

Inside loop:

complement in seen → O(1)

seen.add() → O(1)

pairs.add() → O(1)

(All average-case O(1) because Python set uses hashing.)

Final Time Complexity: O(n)

#### 📦 Space Complexity

We store:
seen set → up to n elements

pairs set → worst case up to n/2 pairs

So total storage proportional to n.


✅ Space Complexity:
O(n)



----

#### Direct Comparison

| Approach            | Time  | Space       |
| ------------------- | ----- | ----------- |
| Brute Force         | O(n²) | O(n²) worst |
| Optimized (Hashing) | O(n)  | O(n)        |


#### 🎯 Interview Explanation (Short Version)

If interviewer asks:

Why is optimized O(n)?

You say:

Because we traverse the array once and use constant-time hash lookups to check complements.

If interviewer asks:

Why brute force is O(n²)?

You say:

Because we compare every pair of elements using nested loops.
